```python
import osmnx as ox
import geopandas as gpd
import matplotlib.pyplot as plt
```

```python
# ======================================================
# Configuración
# ======================================================

# Centro aproximado de Medellín (Parque Berrío)
lat = 6.25184
lon = -75.56359

# Radio en metros
radius = 30000   # 30 km

# Tipo de red
network_type = "drive"
# Opciones:
# "drive"
# "walk"
# "bike"
# "all"
# "all_public"
```

```python
# ======================================================
# Descargar red vial
# ======================================================

print("Descargando red vial...")

G = ox.graph_from_point(
    (lat, lon),
    dist=radius,
    network_type=network_type,
    simplify=True
)

print("Red descargada.")
```

```python
# ======================================================
# Visualización de la red con OSMnx
# ======================================================

fig, ax = ox.plot_graph(
    G,
    figsize=(20,20),
    bgcolor="white",
    node_size=0,
    edge_color="gray",
    edge_linewidth=0.4,
    show=False,
    close=False
)

ax.set_title("Red vial OSM - Medellín (30 km)")

plt.show()
```

```python
# ======================================================
# Convertir a GeoDataFrames
# ======================================================

nodes, edges = ox.graph_to_gdfs(G)

print(f"Nodos: {len(nodes)}")
print(f"Aristas: {len(edges)}")

# ======================================================
# Crear carpeta de salida
# ======================================================

output_folder = "OSM_Medellin_30km"

import os
os.makedirs(output_folder, exist_ok=True)

# ======================================================
# Exportar Shapefiles
# ======================================================

nodes.to_file(f"{output_folder}/MDE_nodes.shp")

edges.to_file(f"{output_folder}/MDE_edges.shp")

print("Archivos exportados correctamente.")
```
